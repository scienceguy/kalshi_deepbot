# Ollama on Kaggle

Install and run [Ollama](https://ollama.com) in a Kaggle notebook.

**Before running:** in notebook Settings enable **Internet** (and **GPU** if you want to run larger models).

````python
# Install Ollama
!curl -fsSL https://ollama.com/install.sh | sh
````

````python
# Start Ollama server in the background
import subprocess, time

server = subprocess.Popen(["ollama", "serve"], stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
time.sleep(5)
print("Ollama server pid:", server.pid)
````

````python
# Verify installation
!ollama --version
!ollama list
````
